```python
import numpy as np
import matplotlib.pyplot as plt

def lcg(m, a, c, Z0, n):
    """Mengembalikan (list Z, list U) sesuai Zi = (a*Z(i-1) + c) mod m."""
    Z = [Z0]
    U = []
    for i in range(n):
        Zi = (a * Z[-1] + c) % m
        Z.append(Zi)
        U.append(Zi / m)
    return Z, U

# Parameter LCG kecil untuk diuji (sama dengan Bagian A)
m, a, c, Z0 = 10, 11, 3, 5
Z, U = lcg(m, a, c, Z0, n=30)
print("Barisan Z:")
print(Z)

# Temukan periode: cari di indeks ke berapa Z0 muncul lagi
period = None
for i in range(1, len(Z)):
    if Z[i] == Z[0]:
        period = i
        break

print("Periode =", period)
if period == m:
    print("Full Period")
else:
    print("Bukan Full Period")
```

```python
SEED = 3224012  # seed tetap: NIM saya

# LCG dengan m besar (full period, sudah diuji)
m2, a2, c2, Z0_2 = 2048, 1229, 1, 7
_, U_lcg = lcg(m2, a2, c2, Z0_2, n=1000)
U_lcg = np.array(U_lcg)

# Pembanding: default_rng()
rng = np.random.default_rng(seed=SEED)
U_rng = rng.random(1000)

print("Rata-rata -> LCG:", U_lcg.mean(), " | default_rng:", U_rng.mean())
print("Varians -> LCG:", U_lcg.var(), " | default_rng:", U_rng.var())

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].hist(U_lcg, bins=20); axes[0].set_title("LCG buatan sendiri")
axes[1].hist(U_rng, bins=20); axes[1].set_title("default_rng()")
plt.show()
```

```python
beta = 3  # menit, mean distribusi target
n_sample = 500
U_arrival = rng.random(n_sample)

# Inverse transform manual (JANGAN pakai rng.exponential())
X_arrival = -beta * np.log(U_arrival)

print("Contoh satu pasangan U dan X:")
print("U =", U_arrival[0], " -> X =", X_arrival[0])
```

```python
print("Rata-rata sampel X:", X_arrival.mean(), " (target beta =", beta, ")")

x_vals = np.linspace(0, X_arrival.max(), 200)
pdf_vals = (1 / beta) * np.exp(-x_vals / beta)

plt.hist(X_arrival, bins=30, density=True, alpha=0.6, label="Sampel Anda")
plt.plot(x_vals, pdf_vals, "r-", label="PDF eksponensial target")
plt.legend()
plt.show()
```

```python
def run_pipeline(seed, beta=3, n=500):
    rng_local = np.random.default_rng(seed=seed)
    U = rng_local.random(n)
    X = -beta * np.log(U)
    return X

seed_a = 3224012   # NIM saya
seed_b = 3224012   # sengaja SAMA
seed_c = 3224013   # HARUS BEDA

X_run1 = run_pipeline(seed_a)
X_run2 = run_pipeline(seed_b)
X_run3 = run_pipeline(seed_c)

print("Run1 identik dengan Run2 (seed sama)?", np.array_equal(X_run1, X_run2))
print("Run1 identik dengan Run3 (seed beda)?", np.array_equal(X_run1, X_run3))
```
